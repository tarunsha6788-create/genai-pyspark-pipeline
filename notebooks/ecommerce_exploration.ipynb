# E-commerce Sales Analytics with PySpark

Interactive analytics using the `SalesAnalytics` engine with Adaptive Query Execution and Kryo serialization.

```python
from src.config import ORDERS_PARQUET, PRODUCTS_PARQUET, CUSTOMERS_PARQUET
from src.spark_analytics import SalesAnalytics

# Initialize SalesAnalytics (automatically configures 4GB SparkSession, AQE, and Kryo)
analytics = SalesAnalytics()

# Load Parquet datasets
orders_df = analytics.load_parquet(ORDERS_PARQUET)
products_df = analytics.load_parquet(PRODUCTS_PARQUET)
customers_df = analytics.load_parquet(CUSTOMERS_PARQUET)
```

```python
# Top 10 VIP Customers by Total Spend
top_cust_df = analytics.top_customers_by_revenue(orders_df, products_df, n=10, customers_df=customers_df)
top_cust_df.show(truncate=False)
```

```python
# Sales and Revenue Aggregations by Product Category
category_df = analytics.sales_by_category(orders_df, products_df)
category_df.show(truncate=False)
```

```python
# Month-over-Month Revenue Growth Trends using PySpark Window Functions
trends_df = analytics.monthly_trends(orders_df, products_df)
trends_df.show(24, truncate=False)
```
